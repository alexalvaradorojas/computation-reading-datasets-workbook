## 📈 Exploratory Data Analysis: Missing and Duplicate Observations

We will take a closer look at how much information is missing or "mischaracterized" in our Meta-Kaggle dataset, and whether there are any duplicate dataset observations.

```python
# Check missing values for each column
print("Missing Values per Column:")
print("=" * 50)
missing_values = main_df.isnull().sum()
missing_percentage = (main_df.isnull().sum() / len(main_df)) * 100

# Create a summary dataframe
missing_summary = pd.DataFrame({
    'Missing_Count': missing_values,
    'Missing_Percentage': missing_percentage
})

# Sort by missing count (descending) and display
missing_summary = missing_summary.sort_values('Missing_Count', ascending=False)
print(missing_summary)

print("\n" + "=" * 50)
print(f"Total columns with missing values: {(missing_values > 0).sum()}")
```

💡 Key observations:

*   Few datasets are published by organizations
*   Most datasets are not part of competitions
*   Most datasets are missing a subtitle and description
*   Some datasets were generated before tracking ownership
*   Some dasets are missing a title, license, and version ID
*   All datasets have some level of interaction (views, votes, downloads, kernels/projects)



❓ How might we visualize these "silences" or missing information?

```python
# "Waffle Chart"
# Filter for columns with missing data and sort
missing_only = missing_summary[missing_summary['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

# Create waffle charts
fig, axes = plt.subplots(2, 6, figsize=(18, 6))
axes = axes.flatten()

for idx, (col_name, row) in enumerate(missing_only.iterrows()):
    # Create 10x10 grid
    grid = np.zeros((10, 10))
    flat_grid = grid.flatten()
    present_cells = 100 - int(row['Missing_Percentage'])
    flat_grid[:present_cells] = 1
    np.random.shuffle(flat_grid)
    grid = flat_grid.reshape((10, 10))

    # Plot
    sns.heatmap(grid, cmap=['#bdc3c7', '#27ae60'], cbar=False, square=True,
                linewidths=1, linecolor='white', ax=axes[idx],
                xticklabels=False, yticklabels=False)
    axes[idx].set_title(f"{col_name}\n{row['Missing_Percentage']:.2f}% missing",
                        fontsize=9, fontweight='bold')

# Hide unused subplots
for i in range(len(missing_only), len(axes)):
    axes[i].axis('off')

fig.suptitle('Missing Data Visualization\n(Each square = 1% of dataset)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 📈 Measures of Central Tendency (mean, mode, and median)

`📝 These measures help us make sense of the more typical values or where the  observations center`

* *Mean* refers to the average pattern of our observations, but it is sensitive to "outliers" (very low or high-scoring values).

* *Median* refers to the "middle point" in our sorted observations, it is resistant to "outliers".

* *Mode* refers to most frequent n value in our observations.  

Each measure tells a different story or perspective on our data.

```python
# Calculate the mean, mode, and median of 'TotalViews'
mean_val = main_no_missing_df['TotalViews'].mean()
median_val = main_no_missing_df['TotalViews'].median()
mode_val = main_no_missing_df['TotalViews'].mode().iloc[0]

print("Mean: {:.2f}".format(mean_val))
print("Median: {:.2f}".format(median_val))
print("Mode: {:.2f}".format(mode_val))
```

💡 Key observations:
*   On average, a dataset gets 1,901 views
*   The "half" point of views received is 146
*   The most common number of views received is 8

## 📈 Measures of Dispersion/Spread (standard deviation, IQR range)

`📝 These measures help us make sense of the dispersion or spread of our observations`

* *Range* refers to the "span" of your observations by subtracting the highest (maximum) value from the lowest (minimum) value

* *Standard deviation* (std) refers average distance of all values from the mean, how much the observations "wiggle" around the mean

* *Interquartile* range (IQR) refers to the spread of central values or the middle 50% of all values

These measures help make sense of how reliable or meaningful our central measures for interpreting the observations (mean, mode, median)

```python
# Calculate the range of 'TotalViews'
views_range = main_df['TotalViews'].max() - main_df['TotalViews'].min()
print(f"Range of TotalViews: {views_range:,.0f}")
print(f"Minimum: {main_df['TotalViews'].min():,.0f}")
print(f"Maximum: {main_df['TotalViews'].max():,.0f}")
```

💡 Key observations:
*  There is at least one dataset with 0 views (minumum)
*  There is at least one dataset with 12,089,246 views (maximum)
*  The range or span of views is between 0 and 12,089,246

### 📈📊 Correlations and Scatterplots: Visualize Relationship Between Total Views and Total Kernels

`📝 A correlation measures the strength and direction of the relationship of two variables`

Correlations help us identify whether and how two variables move together. Two key concepts to help us interpet them are:

**Direction**
*  **Positive (+):** As one variable increases, the other also increases
*  **Negative (-):** As one variable decreases, the other increases
*  **No correlation (0):** There is norelationship between variables

**Strength** (ranges from 1.0 to -1.0)
* +1: Perfect positive correlation
* 0.5: Moderate positive correlation
* 0: No correlation
* -0.5: Moderate negative correlation
* -1: Perfect negative correlation

⚠️ There are many correlations that are used depending on the properties of the data (e.g., linear and non-linear relationships; only numerical and continuous variables).

⚠️ "Correlation does not cause causation". Knowing that two variables are correlated does not mean there is an effect or influence of one variable on another variable.

📊 A common way to determine if our variables are linear or non-linear is by visualizing their relationship using a scatterplot.    Let's use a scatterplot to visualize the relationship bewteen Total Views and Total Kernels, including the log-transformed versions of these variables to compare.

```python
# Create side-by-side plots
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Original scale (left plot)
sns.regplot(x='TotalViews',
            y='TotalKernels',
            data=main_no_missing_df,
            scatter_kws={'alpha':0.5, 'edgecolors':'k'},
            line_kws={'color':'red', 'linewidth':2},
            ax=axes[0])

axes[0].set_xlabel('Total Views', fontsize=12)
axes[0].set_ylabel('Total Kernels', fontsize=12)
axes[0].set_title('Original Scale', fontsize=13, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Log-transformed scale (right plot)
# Filter out zeros to avoid log(0) issues
data_no_zeros = main_no_missing_df[(main_no_missing_df['TotalViews'] > 0) &
                                     (main_no_missing_df['TotalKernels'] > 0)].copy()

# Create log-transformed columns
data_no_zeros['LogTotalViews'] = np.log10(data_no_zeros['TotalViews'])
data_no_zeros['LogTotalKernels'] = np.log10(data_no_zeros['TotalKernels'])

sns.regplot(x='LogTotalViews',
            y='LogTotalKernels',
            data=data_no_zeros,
            scatter_kws={'alpha':0.5, 'edgecolors':'k'},
            line_kws={'color':'red', 'linewidth':2},
            ax=axes[1])

axes[1].set_xlabel('Log10(Total Views)', fontsize=12)
axes[1].set_ylabel('Log10(Total Kernels)', fontsize=12)
axes[1].set_title('Log-Transformed Scale', fontsize=13, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.suptitle('Relationship between Total Views and Total Kernels',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

💡 Key observation:

*  Our variables do not follow a strict linear pattern; they are clustered around lower values.
*  The relationship between Total Views and Total Kernels is non-linear
*  Visualizing the log-transformed variables shows us a different scale of the observations as if it is "zooming in"

⚠️ We will use a Spearman correlation to come up with a measure of the strength and direction of the relationship between "Total Views' and "Total Kernels"

```python
# Calculate Spearman correlation for 'TotalViews' and 'TotalKernels'
spearman_corr = main_no_missing_df['TotalViews'].corr(main_no_missing_df['TotalKernels'], method='spearman')

print(f"Spearman Correlation: {spearman_corr:.4f}")

# Interpretation for Spearman
if abs(spearman_corr) > 0.7:
    strength_spear = "Strong"
elif abs(spearman_corr) > 0.4:
    strength_spear = "Moderate"
elif abs(spearman_corr) > 0.1:
    strength_spear = "Weak"
else:
    strength_spear = "No"

direction_spear = "positive" if spearman_corr > 0 else "negative"

print(f"Interpretation: {strength_spear} {direction_spear} correlation")
```

💡 Key observation:

* There is a weak positive correlation between Total Views and Total Kernels
* This suggests that datasets with more views tend to also have more kernels, and vice-versa
* BUT we cannot claim that more views leads to more kernels

⚠️ There are other types of correlations such as the Pearson correlation that is commonly used with variables that have a linear relationship.

```python
# Calculate Pearson correlation for 'TotalViews' and 'TotalKernels'
pearson_corr = main_no_missing_df['TotalViews'].corr(main_no_missing_df['TotalKernels'], method='pearson')

print(f"Pearson Correlation: {pearson_corr:.4f}")

# Interpretation
if abs(pearson_corr) > 0.7:
    strength = "Strong"
elif abs(pearson_corr) > 0.4:
    strength = "Moderate"
elif abs(pearson_corr) > 0.1:
    strength = "Weak"
else:
    strength = "No"

direction = "positive" if pearson_corr > 0 else "negative"

print(f"Interpretation: {strength} {direction} correlation")
```

💡 Key observation:

* The  0.6 score suggests there is a moderate positive bewteen the number of views a dataset receives and the number of kernels/projects linked to a dataset

⚠️ While both correlation scores (Spearman and Pearson) are positive, the Spearman score "corrects" for the influence of outliers, which may "pull" the observations in a certain direction, distorting the strength of the relationship.

# 2️⃣ Forensic Qualitative Analysis

## 💭 Interpretive Strategies for Reading Datasets

We will review three different interpretive strategies for reading datasets using Forensic Qualitative Analysis.

🕵️ Forensic Qualitative Analysis allows us to `“to imbue digital trace material with meaning through a detailed study
of the trace materials themselves, along with their context and connections with other
events and materials”` [(Champion et al., 2019: 7)]((https://dl.acm.org/doi/pdf/10.1145/3359155))

It is a method to re-construct contexts by tracing different elements that constitute an event or situation, similar to "detective work"

We will complement this appraoch with "Literal, Connotative, and Deconstructive" readings of datasets. In her article, [Poirier, 2021](https://journals.sagepub.com/doi/full/10.1177/20539517211029322#_i22) presents three strategies to reading datasets:

[![Alt text](https://raw.githubusercontent.com/alexalvaradorojas/computation-reading-datasets-workbook/main/image-assets/poirier_2021_strategies.png)](https://raw.githubusercontent.com/alexalvaradorojas/computation-reading-datasets-workbook/main/image-assets/poirier_2021_strategies.png)


### 🔍 Literal Reading



*   How are variables defined in the original codebook or dictionary?
*   How are observations characterized according to categories (e.g., numerical, text, visual, date/time)
*   What information "makesup" these categories and variables?



### 🔍 Connotative Reading



*   How have the variables and observations changed over time?
*   Who created this dataset and for what purpose?
*   What are the specific conditions that prompted the production of this dataset?
*   What other materials help us contextualize the "life" of this dataset?



### 🔍 Deconstructive Reading



*  What are the silences or abscences in the data and why are they important?
*  What are the politics in constructing variables and characterizing observations according to categories (e.g., numerical, text, visual, data/time?)
*  Who is and is not represented and how are these representations made durable?
